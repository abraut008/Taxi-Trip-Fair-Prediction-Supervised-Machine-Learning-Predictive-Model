## Statistical Feature Selection (Categorical Vs Continuous) using ANOVA test

```python
TaxiTripFairData.columns
```

```python
categorical_Columns=['Time_of_Day', 'Day_of_Week', 'Passenger_Count', 'Traffic_Conditions', 'Weather']
from scipy.stats import f_oneway
SelectedPredictors=[]
for cols in categorical_Columns:
    categorcalGroupLists=TaxiTripFairData.groupby(cols)['Trip_Price'].apply(list) 
    AnovaResults=f_oneway(*categorcalGroupLists)
    if AnovaResults[1]<0.05:
        print(cols,' is correlated with ','Trip_Price' ,"and  Pvalue is:",AnovaResults[1])
        SelectedPredictors.append(cols)
    else:
        print(cols,' is not correlated with ','Trip_Price' ,"and  Pvalue is:",AnovaResults[1])
print(SelectedPredictors)
```
